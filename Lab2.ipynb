```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('company_sales_data.csv')

x = df['month_number'].to_numpy()
y = df['total_profit'].to_numpy()

fig = plt.figure()
ax = fig.add_axes([0, 0, 1, 1])

ax.plot(
    x, y, 
    linestyle='dotted', 
    color='red', 
    marker='o', 
    markerfacecolor='red', 
    markeredgecolor='red', 
    linewidth=3, 
    label='Profit data of last year'
)
ax.set_xlabel('Month Number')
ax.set_ylabel('Sold units number')
ax.set_title('Company Sales Data with Custom Styles')

ax.legend(loc='lower right')
```
